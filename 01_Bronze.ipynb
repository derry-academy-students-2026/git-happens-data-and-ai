**Creating the bronze layer for taxi data**

```python
from pyspark.sql.functions import current_timestamp

Catalogue = 'students_data'
Schema = 'git-happens-schema'

taxi_df = spark.read.csv("/Volumes/students_data/diarmuid-gallagher/technologist-volume/taxi_data.csv", header=True, inferSchema=True)

# Replace spaces in column names with underscores
taxi_df = taxi_df.toDF(*[c.replace(" ", "_") for c in taxi_df.columns])

taxi_df = taxi_df.withColumn("timestamp", current_timestamp())

taxi_df.write.mode("overwrite").saveAsTable(f"{Catalogue}.`{Schema}`.bronze_taxi")
```
